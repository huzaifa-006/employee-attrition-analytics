# Employee Attrition Analytics — Business Insights

## Project Overview

This project analyzes employee data to understand employee attrition and identify factors associated with employee turnover.

The analysis combines:

- Data cleaning
- Exploratory Data Analysis (EDA)
- Statistical analysis
- Machine Learning
- Feature importance analysis
- Tableau visualization
- SQL analysis

The primary business objective is to identify employee groups and factors associated with higher attrition so that organizations can develop better employee retention strategies.

## Load Final Employee Dataset

We will load the cleaned employee dataset used throughout the project.

This dataset contains the employee attributes and the original `Attrition` target variable.

```python
import pandas as pd

df = pd.read_csv("../data/processed/employee_hr_analytics_clean.csv")

df.head()
```

## Dataset Overview

The dataset contains employee-level demographic, professional, compensation, workload, satisfaction, and employment information.

The target variable for attrition analysis is:

- `Attrition`

where:

- `Yes` indicates that the employee left the organization.
- `No` indicates that the employee remained with the organization.

```python
print("Number of Employees:", df.shape[0])
print("Number of Features:", df.shape[1])

print("\nDataset Columns:")
print(df.columns.tolist())
```

## Overall Employee Attrition

The overall attrition rate provides a baseline for evaluating employee turnover across different employee groups.

```python
total_employees = len(df)

employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = (employees_left / total_employees) * 100

print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print(f"Overall Attrition Rate: {attrition_rate:.2f}%")
```

## Attrition by Department

Department-level attrition analysis helps identify organizational areas where employee turnover may be relatively high.

```python
department_analysis = (
    df.groupby("Department")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

department_analysis["Attrition_Rate"] = (
    department_analysis["Employees_Left"] / department_analysis["Total_Employees"]
) * 100

department_analysis = department_analysis.sort_values("Attrition_Rate", ascending=False)

department_analysis
```

## Attrition by Job Role

Job-role analysis allows us to identify specific roles where employee turnover may be comparatively high.

These findings can help HR teams investigate role-specific workload, compensation, career progression, or management factors.

```python
job_role_analysis = (
    df.groupby("Job_Role")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

job_role_analysis["Attrition_Rate"] = (
    job_role_analysis["Employees_Left"] / job_role_analysis["Total_Employees"]
) * 100

job_role_analysis = job_role_analysis.sort_values("Attrition_Rate", ascending=False)

job_role_analysis
```

## Attrition by Overtime

Overtime is an important workload-related factor.

We will compare attrition rates between employees who work overtime and employees who do not.

```python
overtime_analysis = (
    df.groupby("Overtime")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

overtime_analysis["Attrition_Rate"] = (
    overtime_analysis["Employees_Left"] / overtime_analysis["Total_Employees"]
) * 100

overtime_analysis
```

## Attrition by Job Satisfaction

Job satisfaction is an important employee-retention indicator.

We will compare attrition rates across different job satisfaction levels.

```python
satisfaction_analysis = (
    df.groupby("Job_Satisfaction")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

satisfaction_analysis["Attrition_Rate"] = (
    satisfaction_analysis["Employees_Left"] / satisfaction_analysis["Total_Employees"]
) * 100

satisfaction_analysis
```

## Attrition by Experience

Employee experience may influence retention patterns.

We will compare average experience between employees who stayed and employees who left.

```python
experience_analysis = (
    df.groupby("Attrition")["Experience"]
    .agg(["count", "mean", "median", "min", "max"])
    .reset_index()
)

experience_analysis
```

## Salary and Attrition

Compensation is another factor that may be associated with employee retention.

We will compare salary statistics between employees who stayed and employees who left.

```python
salary_analysis = (
    df.groupby("Attrition")["Salary"]
    .agg(["count", "mean", "median", "min", "max"])
    .reset_index()
)

salary_analysis
```

## Promotion History and Attrition

Career progression can influence employee retention.

We will compare attrition rates between employees who received a promotion within the last five years and those who did not.

```python
promotion_analysis = (
    df.groupby("Promotion_Last_5Yrs")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

promotion_analysis["Attrition_Rate"] = (
    promotion_analysis["Employees_Left"] / promotion_analysis["Total_Employees"]
) * 100

promotion_analysis
```

## Remote Work and Attrition

We will compare employee attrition across different remote-work arrangements.

```python
remote_work_analysis = (
    df.groupby("Remote_Work")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)
remote_work_analysis["Attrition_Rate"] = (
    remote_work_analysis["Employees_Left"] / remote_work_analysis["Total_Employees"]
) * 100

remote_work_analysis
```

## Key Business Metrics

We will create a compact summary of the organization's overall employee metrics.

```python
summary = pd.DataFrame(
    {
        "Metric": [
            "Total Employees",
            "Employees Who Left",
            "Overall Attrition Rate",
            "Average Salary",
            "Average Experience",
            "Average Job Satisfaction",
            "Average Work Hours",
        ],
        "Value": [
            len(df),
            (df["Attrition"] == "Yes").sum(),
            round((df["Attrition"] == "Yes").mean() * 100, 2),
            round(df["Salary"].mean(), 2),
            round(df["Experience"].mean(), 2),
            round(df["Job_Satisfaction"].mean(), 2),
            round(df["Work_Hours"].mean(), 2)
        ]
    }
)

summary
```

## Identify Highest-Attrition Groups

We will programmatically identify the groups with the highest observed attrition rates.

```python
highest_department = department_analysis.iloc[0]

highest_job_role = job_role_analysis.iloc[0]

highest_overtime = overtime_analysis.sort_values(
    "Attrition_Rate", ascending=False
).iloc[0]

highest_satisfaction = satisfaction_analysis.sort_values(
    "Attrition_Rate", ascending=False
).iloc[0]

print(
    "Highest Department Attrition:",
    highest_department["Department"],
    f"({highest_department['Attrition_Rate']:.2f}%)",
)

print(
    "Highest Job Role Attrition:",
    highest_job_role["Job_Role"],
    f"({highest_job_role['Attrition_Rate']:.2f}%)",
)

print(
    "Highest Overtime Attrition:",
    highest_overtime["Overtime"],
    f"({highest_overtime['Attrition_Rate']:.2f}%)",
)

print(
    "Highest Satisfaction-Level Attrition:",
    highest_satisfaction["Job_Satisfaction"],
    f"({highest_satisfaction['Attrition_Rate']:.2f}%)",
)
```

## Export Business Analysis Results

The main analysis tables will be exported so they can be reused in reports, presentations, or future dashboards.

```python
department_analysis.to_csv(
    "../data/processed/business_department_analysis.csv", index=False
)

job_role_analysis.to_csv(
    "../data/processed/business_job_role_analysis.csv", index=False
)

overtime_analysis.to_csv(
    "../data/processed/business_overtime_analysis.csv", index=False
)

satisfaction_analysis.to_csv(
    "../data/processed/business_satisfaction_analysis.csv", index=False
)

promotion_analysis.to_csv(
    "../data/processed/business_promotion_analysis.csv", index=False
)

remote_work_analysis.to_csv(
    "../data/processed/business_remote_work_analysis.csv", index=False
)

summary.to_csv("../data/processed/business_summary.csv", index=False)

print("Business analysis results exported successfully.")
```
