# Employee Attrition Analytics

## Module 3: Data Cleaning

### Objective
Clean and validate the employee dataset by handling missing values, duplicates, inconsistent categories, invalid values, and outliers.

```python
import pandas as pd
import numpy as np
```

## Load Dirty Dataset

```python
df = pd.read_csv("../data/raw/employee_hr_analytics_dirty.csv")

df.head()
```

## Inspect Dataset Structure

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

## Dataset Information

```python
df.info()
```

## Summary Statistics

```python
df.describe()
```

## Analyze Missing Values

```python
missing_count = df.isnull().sum()

missing_count
```

## Missing Value Percentage

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_percentage = missing_percentage.round(2)

missing_percentage[missing_percentage > 0]
```

## Handle Missing Values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Job_Satisfaction"] = df["Job_Satisfaction"].fillna(df["Job_Satisfaction"].median())

df["Department"] = df["Department"].fillna(df["Department"].mode()[0])
```

## Verify Missing Values

```python
df.isnull().sum()
```

## Detect Duplicate Records

```python
duplicate_count = df.duplicated().sum()

print("Duplicated rows:", duplicate_count)
```

## Remove Duplicate Records

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

## Verify Duplicates

```python
print("Duplicate rows:", df.duplicated().sum())
```

## Convert Joining Date to Datetime

```python
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])

df["Joining_Date"].dtype
```

## Remove Extra Spaces

```python
df["City"] = df["City"].str.strip()
```

## Verify City Values

```python
df["City"].unique()
```

## Standardize Department Names

```python
department_mapping = {
    "I.T": "IT",
    "Human Resource": "HR",
    "finance": "Finance",
    "MARKETING": "Marketing",
}

df["Department"] = df["Department"].replace(department_mapping)
```

## Verify Department Categories

```python
df["Department"].value_counts()
```

## Detect Invalid Ages

```python
invalid_age = df[(df["Age"] < 18) | (df["Age"] > 65)]

invalid_age[["Employee_ID", "Age"]]
```

## Correct Invalid Ages

```python
median_age = df.loc[(df["Age"] >= 18) & (df["Age"] <= 65), "Age"].median()

df.loc[(df["Age"] < 18) | (df["Age"] > 65), "Age"] = median_age
```

## Verify Age Range

```python
print("Minimum age:", df["Age"].min())
print("Maximum age:", df["Age"].max())
```

## Detect Invalid Salaries

```python
invalid_salary = df[df["Salary"] <= 0]

invalid_salary[["Employee_ID", "Salary"]]
```

## Correct Invalid Salaries

```python
valid_salary_median = df.loc[
    df["Salary"] > 0,
    "Salary",
].median()

df.loc[df["Salary"] <= 0, "Salary"] = valid_salary_median
```

## Verify Salary Values

```python
print("Minimun salary:", df["Salary"].min())
print("Maximum slary:", df["Salary"].max())
```

## Detect Salary Outliers Using IQR

```python
Q1 = df["Salary"].quantile(0.25)
Q3 = df["Salary"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3", Q3)
print("IQR", IQR)
print("Lower bound:", lower_bound)
print("Upper bound:", upper_bound)
```

## Count Salary Outliers

```python
salary_outliers = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]
print("Number of salary outliers:", len(salary_outliers))
```

## Inspect Salary Outliers

```python
salary_outliers[
    ["Employee_ID", "Experience", "Department", "Job_Role", "Salary"]
].sort_values("Salary", ascending=False)
```

## Cap Salary Outliers

```python
df["Salary"] = df["Salary"].clip(lower=lower_bound, upper=upper_bound)
```

## Verify Salary Outliers

```python
print("Minimum salary:", df["Salary"].min())
print("Maximum salary:", df["Salary"].max())

remaining_outliers = df[(df["Salary"] < lower_bound) | (df["Salary"] > upper_bound)]

print("Remaining outiers:", len(remaining_outliers))
```

## Final Data Quality Check

```python
print("Dataset shape")
print(df.shape)

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())

print("\nMinimum age:")
print(df["Age"].min())

print("\nMaximum age:")
print(df["Age"].max())

print("\nMinimum salary:")
print(df["Salary"].min())

print("\nDepartment Categories:")
print(df["Department"].unique())
```

## Save Clean Dataset

```python
output_path = "../data/processed/employee_hr_analytics_clean.csv"
df.to_csv(output_path, index=False)
print(f"Clean dataset saved to:{output_path}")
```

```python
clean_df = pd.read_csv(output_path)
print("Shape:", clean_df.shape)

print("\nMissing values:", clean_df.isnull().sum().sum())

print("\nDuplicate rows:", clean_df.duplicated().sum())

clean_df.head()
```
