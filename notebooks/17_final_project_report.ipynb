## 3. Load Final Dataset

The final cleaned employee dataset will be loaded for the project summary.

```python
import pandas as pd

df = pd.read_csv("../data/processed/employee_hr_analytics_clean.csv")

df.head()
```

## 4. Dataset Statistics

We will calculate the key statistics required for the final project report.

```python
total_employees = len(df)

employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = (employees_left / total_employees) * 100

average_salary = df["Salary"].mean()
average_experience = df["Experience"].mean()
average_satisfaction = df["Job_Satisfaction"].mean()
average_work_hours = df["Work_Hours"].mean()

print("Total Employees:", total_employees)
print("Employees Who Left:", employees_left)
print(f"Attrition Rate: {attrition_rate:.2f}%")
print(f"Average Salary: {average_salary:.2f}")
print(f"Average Experience: {average_experience:.2f} years")
print(f"Average Job Satisfaction: {average_satisfaction:.2f}")
print(f"Average Work Hours: {average_work_hours:.2f}")
```

## 5. Department Attrition

We will identify departments with relatively high attrition rates.

```python
department_summary = (
    df.groupby("Department")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
    .reset_index()
)

department_summary["Attrition_Rate"] = (
    department_summary["Employees_Left"]
    / department_summary["Total_Employees"]
) * 100

department_summary = department_summary.sort_values(
    "Attrition_Rate",
    ascending=False
)

department_summary
```

## 6. Overtime and Attrition

Overtime is an important workload-related variable.

We will compare attrition rates between employees who work overtime and those who do not.

```python
overtime_summary = (
    df.groupby("Overtime")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
    .reset_index()
)

overtime_summary["Attrition_Rate"] = (
    overtime_summary["Employees_Left"]
    / overtime_summary["Total_Employees"]
) * 100

overtime_summary
```

## 7. Job Satisfaction and Attrition

Job satisfaction is analyzed to determine whether attrition rates differ across satisfaction levels.

```python
satisfaction_summary = (
    df.groupby("Job_Satisfaction")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=(
            "Attrition",
            lambda x: (x == "Yes").sum()
        )
    )
    .reset_index()
)

satisfaction_summary["Attrition_Rate"] = (
    satisfaction_summary["Employees_Left"]
    / satisfaction_summary["Total_Employees"]
) * 100

satisfaction_summary
```

## 8. Salary Comparison

We will compare salary statistics between employees who stayed and employees who left.

```python
salary_summary = (
    df.groupby("Attrition")["Salary"]
    .agg(
        ["count", "mean", "median", "min", "max"]
    )
    .reset_index()
)

salary_summary
```

## 15. Project Limitations

Several limitations should be considered when interpreting the results.

### Synthetic Dataset

The employee dataset used in this project was generated for educational purposes and therefore does not represent a real organization's workforce.

### Correlation vs. Causation

Observed relationships between employee characteristics and attrition do not necessarily imply causal relationships.

### Sample Size

The dataset contains 1,000 employee records, which is sufficient for project-based analysis but may not represent the complexity of a large organization's workforce.

### Model Limitations

Machine learning predictions depend on the quality and representativeness of the available features.

### HR Decision-Making

Predictions should be used as decision-support information and should not be used as the sole basis for employment decisions.
