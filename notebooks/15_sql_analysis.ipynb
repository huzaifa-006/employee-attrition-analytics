# SQL Analysis

In this module, we will use SQL to analyze the cleaned employee dataset.

The employee data will be loaded into a SQLite database and analyzed using SQL queries.

```python
import pandas as pd
import sqlite3
```

## Load the Cleaned Employee Dataset

We will use the cleaned HR dataset created during the data cleaning stage.

```python
df = pd.read_csv("../data/processed/employee_hr_analytics_clean.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

## Create the SQLite Database

We will create a SQLite database and store the employee dataset in a table called `employees`.

```python
connection = sqlite3.connect("../data/employee_attrition.db")

print("SQLite database connected successfully.")
```

```python
df.to_sql("employees", connection, if_exists="replace", index=False)

print("Employee data loaded into SQL table successfully.")
```

## Verify the SQL Table

Before performing analysis, we will verify that the employee table was created correctly.

```python
query = """
SELECT *
FROM employees
LIMIT 5;
"""

result = pd.read_sql_query(query, connection)

result
```

## Select Specific Columns

Instead of selecting every column, we can select only the columns required for an analysis.

```python
query = """
SELECT
    Employee_ID,
    Department,
    Job_Role,
    Salary,
    Attrition
FROM employees
LIMIT 10;
"""

result = pd.read_sql_query(query, connection)

result
```

## Filter Employees by Department

The `WHERE` clause allows us to filter records based on a condition.

We will retrieve employees who work in the IT department.

```python
query = """
SELECT
    Employee_ID,
    Department,
    Job_Role,
    Salary,
    Attrition
FROM employees
WHERE Department = 'IT';
"""

it_employees = pd.read_sql_query(query, connection)

it_employees.head(10)
```

## Identify Employees Who Left

We will filter the dataset to identify employees whose attrition status is `Yes`.

```python
query = """
SELECT
    Employee_ID,
    Department,
    Job_Role,
    Salary,
    Attrition
FROM employees
WHERE Attrition = 'Yes';
"""

employees_left = pd.read_sql_query(query, connection)

employees_left.head(10)
```

## Count Employees Who Left

We will use the `COUNT()` function to calculate the total number of employees who left the organization.

```python
query = """
SELECT
    COUNT(*) AS Employees_Left
FROM employees
WHERE Attrition = 'Yes';
"""

result = pd.read_sql_query(query, connection)

result
```

## Count Total Employees

We will calculate the total number of employees in the dataset.

```python
query = """
SELECT
    COUNT(*) AS Total_Employees
FROM employees;
"""

result = pd.read_sql_query(query, connection)

result
```

## Calculate Overall Attrition Rate

We will calculate the organization's overall employee attrition rate.

Attrition Rate = Employees Left / Total Employees × 100

```python
query = """
SELECT
    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees;
"""

overall_attrition = pd.read_sql_query(query, connection)

overall_attrition
```

## Employee Distribution by Department

The `GROUP BY` clause allows us to divide employees into groups.

We will calculate the number of employees in each department.

```python
query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees
FROM employees
GROUP BY Department
ORDER BY Total_Employees DESC;
"""

department_employee_count = pd.read_sql_query(query, connection)

department_employee_count
```

## Attrition by Department

We will calculate the total employees and the number of employees who left for each department.

```python
query = """
SELECT
    Department,

    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left

FROM employees
GROUP BY Department
ORDER BY Employees_Left DESC;
"""

department_attrition = pd.read_sql_query(query, connection)

department_attrition
```

## Attrition Rate by Department

We will calculate the attrition rate for each department.

This allows us to compare employee turnover between departments of different sizes.

```python
query = """
SELECT
    Department,

    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees

GROUP BY Department

ORDER BY Attrition_Rate DESC;
"""

department_attrition_rate = pd.read_sql_query(query, connection)

department_attrition_rate
```

## Average Salary by Department

Salary is an important HR metric.

We will calculate the average salary for each department to understand differences in compensation across the organization.

```python
connection = sqlite3.connect("../data/employee_attrition.db")

query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    ROUND(AVG(Salary), 2) AS Average_Salary,
    MIN(Salary) AS Minimum_Salary,
    MAX(Salary) AS Maximum_Salary
FROM employees
GROUP BY Department
ORDER BY Average_Salary DESC;
"""

department_salary = pd.read_sql_query(query, connection)

department_salary
```

## Attrition by Department and Overtime

We will now perform a multi-dimensional analysis.

Instead of grouping employees by only one variable, we will group them by:

- Department
- Overtime

This allows us to investigate whether overtime-related attrition differs across departments.

```python
query = """
SELECT
    Department,
    Overtime,
    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees

GROUP BY
    Department,
    Overtime

ORDER BY
    Department,
    Attrition_Rate DESC;
"""

department_overtime_attrition = pd.read_sql_query(query, connection)

department_overtime_attrition
```

## Attrition by Job Satisfaction Level

We will calculate employee attrition for each job satisfaction level.

This allows us to determine whether lower or higher satisfaction groups have different attrition rates.

```python
query = """
SELECT
    Job_Satisfaction,
    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees

GROUP BY Job_Satisfaction

ORDER BY Job_Satisfaction;
"""

satisfaction_attrition = pd.read_sql_query(query, connection)

satisfaction_attrition
```

## Departments with High Attrition

The `HAVING` clause is used to filter grouped results.

We will identify departments where the attrition rate is greater than 12%.

```python
query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees

GROUP BY Department

HAVING Attrition_Rate > 12

ORDER BY Attrition_Rate DESC;
"""

high_attrition_departments = pd.read_sql_query(query, connection)

high_attrition_departments
```

## Departments with More Than 100 Employees

We will use `HAVING` with `COUNT()` to identify departments containing more than 100 employees.

```python
query = """
SELECT
    Department,
    COUNT(*) AS Total_Employees,
    ROUND(AVG(Salary), 2) AS Average_Salary
FROM employees
GROUP BY Department
HAVING COUNT(*) > 100
ORDER BY Total_Employees DESC;
"""

large_departments = pd.read_sql_query(query, connection)

large_departments
```

## Employees with High Salary and Attrition

We will identify employees who left the organization and had salaries above 150,000.

This demonstrates how multiple conditions can be combined using `WHERE`.

```python
query = """
SELECT
    Employee_ID,
    Department,
    Job_Role,
    Experience,
    Salary,
    Attrition
FROM employees
WHERE Attrition = 'Yes'
  AND Salary > 150000
ORDER BY Salary DESC;
"""

high_salary_attrition = pd.read_sql_query(query, connection)

high_salary_attrition.head(20)
```

## Employees with High Workload

We will identify employees who work more than 45 hours per week and determine their attrition status.

This can help investigate the relationship between workload and employee turnover.

```python
query = """
SELECT
    Employee_ID,
    Department,
    Job_Role,
    Work_Hours,
    Overtime,
    Attrition
FROM employees
WHERE Work_Hours > 45
ORDER BY Work_Hours DESC;
"""

high_workload_employees = pd.read_sql_query(query, connection)

high_workload_employees.head(20)
```

## Final SQL Analysis Summary

We will create a summary table containing key organizational metrics.

The metrics include:

- Total employees
- Employees who left
- Average salary
- Average experience
- Average job satisfaction
- Overall attrition rate

```python
query = """
SELECT
    COUNT(*) AS Total_Employees,

    SUM(
        CASE
            WHEN Attrition = 'Yes' THEN 1
            ELSE 0
        END
    ) AS Employees_Left,

    ROUND(AVG(Salary), 2) AS Average_Salary,

    ROUND(AVG(Experience), 2) AS Average_Experience,

    ROUND(AVG(Job_Satisfaction), 2) AS Average_Job_Satisfaction,

    ROUND(
        100.0 *
        SUM(
            CASE
                WHEN Attrition = 'Yes' THEN 1
                ELSE 0
            END
        ) / COUNT(*),
        2
    ) AS Attrition_Rate

FROM employees;
"""

sql_summary = pd.read_sql_query(query, connection)

sql_summary
```
