## Module 11: Streamlit Prediction Application

### Objective

Build an interactive application that uses the trained machine learning model to predict employee attrition risk.

```python
import pandas as pd
import numpy as np
import joblib
import os
```

## Load Saved Model and Feature Schema

```python
model_path = "../models/employee_attrition_model.pkl"
features_path = "../models/employee_attrition_features.pkl"

model = joblib.load(model_path)

feature_columns = joblib.load(features_path)

print("Model loaded successfully.")
print("Number of features:", len(feature_columns))
```

## Model Feature Requirements

```python
for index, feature in enumerate(feature_columns, start=1):
    print(f"{index}. {feature}")
```

## Create Prediction Function

```python
def predict_attrition(employee_data):

    employee_df = pd.DataFrame([employee_data])

    employee_df = employee_df[feature_columns]

    prediction = model.predict(employee_df)[0]

    probability = model.predict_proba(employee_df)[0, 1]

    if probability < 0.30:
        risk = "Low"

    elif probability < 0.60:
        risk = "Medium"

    else:
        risk = "High"

    return prediction, probability, risk
```

## Test Prediction Function

```python
sample_employee = {}

for feature in feature_columns:
    sample_employee[feature] = 0
```

```python
prediction, probability, risk = predict_attrition(sample_employee)

print("Prediction:", prediction)

print("Probability:", round(probability, 4))

print("Risk:", risk)
```
