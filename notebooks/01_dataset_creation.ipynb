# Employee Attrition Analytics

## Module 1: Dataset Creation

### Objective
Create a realistic synthetic employee dataset that will be used throughout this project for data cleaning, exploratory data analysis, feature engineering, machine learning, and dashboard creation.

```python
import pandas as pd
import numpy as np
from faker import Faker
import random
```

## Initialize Faker and Set Random Seeds

Setting random seeds ensures the dataset is reproducible. Every time this notebook is executed, the same dataset will be generated.

```python
fake = Faker()

random.seed(42)
np.random.seed(42)
Faker.seed(42)
```

## Define Dataset Categories

Define all categorical values in one place to make the dataset easier to maintain and update.

```python
departments = ["IT", "HR", "Finance", "Sales", "Marketing"]

job_roles = {
    "IT": ["Software Engineer", "Data Analyst", "DevOps Engineer"],
    "HR": ["HR Executive", "Recruiter"],
    "Finance": ["Accountant", "Financial Analyst"],
    "Sales": ["Sales Executive", "Sales Manager"],
    "Marketing": ["Marketing Specialist", "SEO Executive"],
}

education_levels = ["Bachelor", "Master", "PhD"]

cities = ["Karachi", "Lahore", "Islamabad", "Rawalpindi", "Faisalabad", "Peshawar"]

employment_types = ["Full-Time", "Contract"]
```

## Verify Categories

Display the available categories before generating the dataset.

```python
print("Departments:")
print(departments)

print("\nEducation Levels:")
print(education_levels)

print("\nCities:")
print(cities)
```

## Create an Empty List

Each generated employee record will be stored as a dictionary inside a list. Later, the list will be converted into a Pandas DataFrame.

```python
employees = []

print(type(employees))
```

## Define Data Generation Rules

To make the dataset realistic, several columns will have logical relationships. For example, salary will depend on experience, and job satisfaction will influence attrition.

```python
salary_rules = {
    "0-2": (40000, 70000),
    "3-5": (70000, 100000),
    "6-10": (100000, 150000),
    "10+": (150000, 250000),
}

project_rules = {"0-2": (1, 3), "3-5": (3, 6), "6-10": (6, 10), "10+": (8, 15)}

print("Salary Rules")
print(salary_rules)

print("\nProject Rules")
print(project_rules)
```

## Generate Employee Records

Generate 1,000 realistic employee records using predefined business rules and store them in a list.

```python
employees = []

for employee_id in range(1, 1001):

    # Basic Information
    department = random.choice(departments)
    role = random.choice(job_roles[department])

    age = random.randint(22, 60)
    gender = random.choice(["Male", "Female"])
    experience = random.randint(0, age - 21)

    education = random.choice(education_levels)
    city = random.choice(cities)
    employment = random.choice(employment_types)

    # Salary and Projects
    if experience <= 2:
        salary = random.randint(40000, 70000)
        projects = random.randint(1, 3)

    elif experience <= 5:
        salary = random.randint(70000, 100000)
        projects = random.randint(3, 6)

    elif experience <= 10:
        salary = random.randint(100000, 150000)
        projects = random.randint(6, 10)

    else:
        salary = random.randint(150000, 250000)
        projects = random.randint(8, 15)

    # Performance
    performance = random.randint(1, 5)
    manager_rating = random.randint(4, 10)

    work_hours = random.randint(35, 55)

    overtime = random.choice(["Yes", "No"])
    remote_work = random.choice(["Yes", "No"])

    training_hours = random.randint(5, 80)

    sick_leaves = random.randint(0, 15)

    # Job Satisfaction
    if overtime == "Yes":
        job_satisfaction = random.randint(3, 8)
    else:
        job_satisfaction = random.randint(6, 10)

    # Promotion
    if performance >= 4:
        promotion = random.choice(["Yes", "No"])
    else:
        promotion = "No"

    # Attrition
    if job_satisfaction <= 4:
        attrition = random.choice(["Yes", "Yes", "No"])
    else:
        attrition = "No"

    # Joining Date
    joining_date = fake.date_between(
        start_date="-15y",
        end_date="today"
    )

    # Employee Record
    employee = {
        "Employee_ID": employee_id,
        "Age": age,
        "Gender": gender,
        "Department": department,
        "Job_Role": role,
        "Experience": experience,
        "Education": education,
        "Salary": salary,
        "Performance_Score": performance,
        "Job_Satisfaction": job_satisfaction,
        "Overtime": overtime,
        "Work_Hours": work_hours,
        "Remote_Work": remote_work,
        "Training_Hours": training_hours,
        "Projects": projects,
        "Promotion_Last_5Yrs": promotion,
        "Joining_Date": joining_date,
        "Attrition": attrition,
        "City": city,
        "Employment_Type": employment,
        "Manager_Rating": manager_rating,
        "Sick_Leaves": sick_leaves
    }

    employees.append(employee)

print(f"Total Employees Generated: {len(employees)}")
```

## Create DataFrame

```python
df = pd.DataFrame(employees)

df.head()
```

## Check Dataset Information

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

## Save Dataset

```python
df.to_csv("../data/raw/employee_hr_analytics.csv", index=False)

print("Dataset saved successfully.")
```
