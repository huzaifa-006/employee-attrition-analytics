# Module 14 — Tableau Dashboard Preparation

```python
import pandas as pd
import os

print("Libraries imported successfully.")
```

```python
file_path = "../data/processed/employee_attrition_dashboard.csv"

df = pd.read_csv(file_path)

print("Dataset shape:", df.shape)

df.head()
```

```python
print("Columns:")

for column in df.columns:
    print(column)
```

## Tableau Dataset Validation

```python
print("Missing values:")
print(df.isnull().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print("Attrition values:")
print(df["Attrition"].value_counts())

print("\nDepartments:")
print(df["Department"].value_counts())

print("\nEmployment Types:")
print(df["Employment_Type"].value_counts())
```

## Create HR KPI Dataset

```python
total_employees = len(df)

employees_left = (df["Attrition"] == "Yes").sum()

attrition_rate = (employees_left / total_employees) * 100

average_salary = df["Salary"].mean()

average_satisfaction = df["Job_Satisfaction"].mean()

average_experience = df["Experience"].mean()

average_work_hours = df["Work_Hours"].mean()

average_training_hours = df["Training_Hours"].mean()

kpi_data = {
    "Total Employees": total_employees,
    "Employees Left": employees_left,
    "Attrition Rate": round(attrition_rate, 2),
    "Average Salary": round(average_salary, 2),
    "Average Job Satisfaction": round(average_satisfaction, 2),
    "Average Experience": round(average_experience, 2),
    "Average Work Hours": round(average_work_hours, 2),
    "Average Training Hours": round(average_training_hours, 2),
}

kpi_data
```

```python
kpi_df = pd.DataFrame(list(kpi_data.items()), columns=["KPI", "Value"])

kpi_df
```

```python
kpi_df.to_csv("../data/processed/hr_dashboard_kpis.csv", index=False)

print("KPI dataset saved successfully.")
```

## Department Attrition Dataset

```python
department_df = (
    df.groupby("Department")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

department_df["Attrition_Rate"] = (
    department_df["Employees_Left"] / department_df["Total_Employees"]
) * 100

department_df
```

```python
department_df.to_csv("../data/processed/tableau_department_analysis.csv", index=False)

print("Department analysis saved.")
```

## Job Role Attrition Dataset

```python
job_role_df = (
    df.groupby("Job_Role")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

job_role_df["Attrition_Rate"] = (
    job_role_df["Employees_Left"] / job_role_df["Total_Employees"]
) * 100

job_role_df = job_role_df.sort_values("Attrition_Rate", ascending=False)

job_role_df
```

```python
job_role_df.to_csv("../data/processed/tableau_job_role_analysis.csv", index=False)

print("Job role analysis saved.")
```

## Overtime Attrition Dataset

```python
overtime_df = (
    df.groupby("Overtime")
    .agg(
        Total_Employees=("Employee_ID", "count"),
        Employees_Left=("Attrition", lambda x: (x == "Yes").sum()),
    )
    .reset_index()
)

overtime_df["Attrition_Rate"] = (
    overtime_df["Employees_Left"] / overtime_df["Total_Employees"]
) * 100

overtime_df
```

```python
overtime_df.to_csv("../data/processed/tableau_overtime_analysis.csv", index=False)

print("Overtime analysis saved.")
```
