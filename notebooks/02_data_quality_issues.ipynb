# Module 2: Data Quality Issues & Data Cleaning

## Objective

Introduce realistic data quality issues into the employee dataset. These issues will be cleaned and analyzed throughout this module.

```python
import pandas as pd
import numpy as np
import random
```

## Load Dataset

```python
df = pd.read_csv("../data/raw/employee_hr_analytics.csv")

df.head()
```

## Check Dataset Shape

```python
print(df.shape)
```

## Introduce Missing Values

```python
np.random.seed(42)

columns = ["Age", "Salary", "Department", "Job_Satisfaction"]

for column in columns:
    random_rows = np.random.choice(df.index, size=20, replace=False)
    df.loc[random_rows, column] = np.nan
```

## Verify Missing Values

```python
df.isnull().sum()
```

## Create Duplicate Rows

```python
duplicates = df.sample(10, random_state=42)

df = pd.concat([df, duplicates], ignore_index=True)
```

## Verify Dataset Shape

```python
print(df.shape)
```

## Create Inconsistent Department Names

```python
department_changes = {
    "IT": "I.T",
    "HR": "Human Resource",
    "Finance": "finance",
    "Marketing": "MARKETING",
}

df["Department"] = df["Department"].replace(department_changes)
```

## Check Unique Departments

```python
df["Department"].unique()
```

## Add Extra Spaces

```python
df.loc[df.sample(20, random_state=10).index, "City"] = " Karachi "
```

## View City Values

```python
df["City"].unique()
```

## Insert Invalid Values

```python
df.loc[5, "Age"] = 150
df.loc[12, "Age"] = 5

df.loc[8, "Salary"] = -50000
```

## View Invalid Records

```python
df.loc[[5, 8, 12]]
```

## Insert Salary Outliers

```python
df.loc[30, "Salary"] = 5000000
df.loc[55, "Salary"] = 3000000
```

## Verify Maximum Salary

```python
df["Salary"].max()
```

## Save Dirty Dataset

```python
df.to_csv("../data/raw/employee_hr_analytics_dirty.csv", index=False)

print("Dirty dataset saved successfully.")
```
